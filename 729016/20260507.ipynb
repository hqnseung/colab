```python
setence = {}
text = input('문자열을 입력하시오:')
for i in text:
    if i in setence:
        setence[i] += 1
    else:
        setence[i] = 1
print(setence)
```

```python
set1 = {10, 20, 30, 40, 50, 60}
set2 = {30, 40, 50, 60, 70, 80}
print(f"set1= {set1}")
print(f"set2= {set2}")
only_one_side = []
for item in set1:
    if item not in set2:
        only_one_side.append(item)
for item in set2:
    if item not in set1:
        only_one_side.append(item)
print(f"어느 한쪽에만 있는 항목들= {set(only_one_side)}")
```
