```python
import math
class Circle:
    def __init__(self, r = 0):
        self.radius = r

    def getArea(self):
        return math.pi * self.radius ** 2

    def getPerimeter(self):
        return math.pi * self.radius * 2
```

```python
c1 = Circle(10)
print(f'반지름 10의 원의 넓이: {c1.getArea():.2f}')
print(f'반지름 10의 원의 둘레: {c1.getPerimeter():.2f}')
c2= Circle(15)
print(f'반지름 15의 원의 넓이: {c2.getArea():.2f}')
print(f'반지름 15의 원의 둘레: {c2.getPerimeter():.2f}')
```

```python
class BankAccount:
    def __init__(self):
        self.__balance = 0

    def withdraw(self, amount):
        self.__balance -= amount
        print(f'계좌에서 {amount}원이 출금되었습니다.')

    def deposit(self, amount):
        self.__balance += amount
        print(f'계좌에서 {amount}원이 입금되었습니다.')
```

```python
a = BankAccount()
a.deposit(50000)
a.withdraw(17000)
```
